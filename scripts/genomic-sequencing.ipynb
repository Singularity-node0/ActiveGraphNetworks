```python
import os
import networkx as nx
import matplotlib.pyplot as plt

# ====================
# Helper Functions
# ====================
def parse_fasta_with_metadata(file_path):
    """
    Parses a .fa file with metadata and yields header, metadata, and sequences.
    """
    header = None
    metadata = {}
    sequence = []
    
    with open(file_path, 'r') as file:
        for line in file:
            line = line.strip()
            if line.startswith('>'):
                if header:
                    yield header, metadata, ''.join(sequence)
                # Extract header and metadata
                header = line.split()[0][1:]  # The first word after '>'
                metadata = {key.split(':')[0]: ':'.join(key.split(':')[1:]) for key in line.split()[1:]}
                sequence = []
            else:
                sequence.append(line)
        # Yield the last sequence
        if header:
            yield header, metadata, ''.join(sequence)

def visualize_graph_with_metadata(graph, title):
    """
    Visualizes the graph with node metadata as labels.
    """
    pos = nx.spring_layout(graph, seed=42)
    plt.figure(figsize=(12, 10))
    node_labels = {
        node: f"{data.get('label', node)}"
        for node, data in graph.nodes(data=True)
    }
    nx.draw(
        graph, pos, with_labels=True, labels=node_labels, node_color='lightblue',
        edge_color='gray', node_size=2000, font_size=8
    )
    edge_labels = nx.get_edge_attributes(graph, 'relationship')
    nx.draw_networkx_edge_labels(graph, pos, edge_labels=edge_labels)
    plt.title(title)
    plt.show()

def import_sequence_with_metadata(graph, sequence_data, sequence_name, metadata):
    """
    Imports a DNA sequence into the graph with metadata.
    """
    graph.add_node(sequence_name, role="sequence", **metadata)
    previous_node = sequence_name
    for i, base in enumerate(sequence_data, 1):
        node_name = f"{sequence_name}_Base_{i}"
        graph.add_node(node_name, base=base, position=i)
        graph.add_edge(previous_node, node_name, relationship="sequence_order")
        previous_node = node_name

def add_synthetic_relationships(graph):
    """
    Adds synthetic relationships to the graph by inferring connections between nodes
    based on metadata and sequence structure.
    """
    # Example 1: Connect sequences on the same chromosome
    sequence_nodes = [
        (node, data) for node, data in graph.nodes(data=True) if data.get("role") == "sequence"
    ]
    for i, (node1, data1) in enumerate(sequence_nodes):
        for node2, data2 in sequence_nodes[i + 1:]:
            if data1.get("chromosome") == data2.get("chromosome"):
                graph.add_edge(node1, node2, relationship="shared_chromosome")

    # Example 2: Link sequences with the same gene_symbol
    for i, (node1, data1) in enumerate(sequence_nodes):
        for node2, data2 in sequence_nodes[i + 1:]:
            if data1.get("gene_symbol") == data2.get("gene_symbol"):
                graph.add_edge(node1, node2, relationship="coexpressed_gene")

    print("Synthetic relationships added!")

# ====================
# Main Execution
# ====================

# Initialize the graph
genome_graph = nx.DiGraph()

# Load and parse the .fa file
fasta_file_path = "/kaggle/input/dna-sequence-dataset/example_dna.fa"

for header, metadata, sequence in parse_fasta_with_metadata(fasta_file_path):
    print(f"Processing: {header}")
    print(f"Metadata: {metadata}")
    import_sequence_with_metadata(genome_graph, sequence, sequence_name=header, metadata=metadata)

# Add synthetic relationships
add_synthetic_relationships(genome_graph)

# Visualize the graph with synthetic relationships
visualize_graph_with_metadata(genome_graph, "Genomic Relationships with Synthetic Edges")
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import networkx as nx
import random

def visualize_3d_graph(graph, title="3D Graph Visualization"):
    """
    Visualizes a graph in 3D space using matplotlib.
    Nodes and edges are plotted in a 3D cube.
    
    Parameters:
        graph (nx.Graph): The graph to visualize.
        title (str): Title of the graph visualization.
    """
    # Create a 3D plot
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    # Generate random 3D positions for nodes
    pos = {node: (random.uniform(-1, 1), random.uniform(-1, 1), random.uniform(-1, 1)) for node in graph.nodes()}
    
    # Plot edges
    for edge in graph.edges(data=True):
        x = [pos[edge[0]][0], pos[edge[1]][0]]
        y = [pos[edge[0]][1], pos[edge[1]][1]]
        z = [pos[edge[0]][2], pos[edge[1]][2]]
        ax.plot(x, y, z, color="gray", alpha=0.7)
    
    # Plot nodes
    for node, coords in pos.items():
        ax.scatter(*coords, s=100, label=node, depthshade=True)
        ax.text(*coords, f"{node}", fontsize=8)
    
    # Set labels
    ax.set_title(title, fontsize=15)
    ax.set_xlabel("X-axis")
    ax.set_ylabel("Y-axis")
    ax.set_zlabel("Z-axis")
    
    plt.show()

# Example graph
example_graph = nx.Graph()
example_graph.add_node("A", role="sequence")
example_graph.add_node("T", role="sequence")
example_graph.add_node("G", role="sequence")
example_graph.add_edge("A", "T", relationship="complementary")
example_graph.add_edge("A", "G", relationship="synthetic")
example_graph.add_edge("T", "G", relationship="sequence_order")

# Visualize the example graph
visualize_3d_graph(example_graph, title="3D Visualization of DNA Relationships")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

def generate_wave(sequence, decay_rate=0.02, frequency=2, amplitude=1):
    """
    Generates a sine wave based on the sequence with optional decay.
    """
    time = np.linspace(0, len(sequence), 1000)  # Time resolution
    wave = np.zeros_like(time)
    
    # Generate a waveform for each base
    for i, base in enumerate(sequence):
        if base == 'A':
            wave += amplitude * np.sin(2 * np.pi * frequency * (time - i)) * np.exp(-decay_rate * (time - i))
        elif base == 'T':
            wave += amplitude * np.cos(2 * np.pi * frequency * (time - i)) * np.exp(-decay_rate * (time - i))
        elif base == 'G':
            wave += amplitude * np.sin(2 * np.pi * (frequency + 1) * (time - i)) * np.exp(-decay_rate * (time - i))
        elif base == 'C':
            wave += amplitude * np.cos(2 * np.pi * (frequency + 1) * (time - i)) * np.exp(-decay_rate * (time - i))
    
    return time, wave

def visualize_dna_wave(sequence, title="DNA Waveform Visualization"):
    """
    Visualizes the DNA sequence as a dynamic waveform in 3D space.
    """
    time, wave = generate_wave(sequence)
    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Create the 3D waveform
    z = wave
    x = time
    y = np.linspace(0, 1, len(z))  # Adding a second dimension for "depth"

    ax.plot(x, y, z, label="DNA Waveform", color="purple", linewidth=2)
    ax.set_xlabel("Time (X Axis)")
    ax.set_ylabel("Depth (Y Axis)")
    ax.set_zlabel("Amplitude (Z Axis)")
    ax.set_title(title)

    plt.show()

# Example usage with a DNA sequence
example_sequence = "ATGCTAGATCG"
visualize_dna_wave(example_sequence)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

def visualize_4d_structure(graph, temporal_key='temporal', layer_gap=10):
    """
    Visualizes the DNA structure with temporal layers in 3D space.
    
    Args:
        graph (nx.Graph): The DNA graph with temporal attributes.
        temporal_key (str): The key used to identify temporal data.
        layer_gap (int): The distance between layers in the Z-axis.
    """
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')

    # Initialize coordinates and colors
    x_coords = []
    y_coords = []
    z_coords = []
    colors = []
    sizes = []

    temporal_layers = {}  # To group nodes by temporal layers

    # Extract nodes and their temporal attributes
    for node, data in graph.nodes(data=True):
        if temporal_key in data:
            time = data[temporal_key]
            if time not in temporal_layers:
                temporal_layers[time] = []
            temporal_layers[time].append(node)

    # Assign positions to each layer
    for z_index, (time, nodes) in enumerate(sorted(temporal_layers.items()), start=1):
        for idx, node in enumerate(nodes):
            x_coords.append(idx % 10)  # Arrange nodes in a grid for each layer
            y_coords.append(idx // 10)
            z_coords.append(z_index * layer_gap)
            colors.append(z_index)  # Color based on layer
            sizes.append(50)  # Uniform size for now

    # Scatter plot for the nodes
    sc = ax.scatter(x_coords, y_coords, z_coords, c=colors, cmap='viridis', s=sizes, alpha=0.8)
    plt.colorbar(sc, label="Temporal Layer")

    # Set labels
    ax.set_title("4D Visualization of DNA Structure with Temporal Layers")
    ax.set_xlabel("X Axis")
    ax.set_ylabel("Y Axis")
    ax.set_zlabel("Temporal Layers (Z Axis)")

    plt.show()

# Example usage
# This assumes you have a graph with 'temporal' attributes already added to nodes
visualize_4d_structure(genome_graph)
```

```python
import plotly.graph_objs as go

def visualize_interactive_3d(graph):
    pos = nx.spring_layout(graph, dim=3, seed=42)
    x, y, z = zip(*[pos[node] for node in graph.nodes])
    edge_x, edge_y, edge_z = [], [], []

    for edge in graph.edges():
        edge_x += [pos[edge[0]][0], pos[edge[1]][0], None]
        edge_y += [pos[edge[0]][1], pos[edge[1]][1], None]
        edge_z += [pos[edge[0]][2], pos[edge[1]][2], None]

    edge_trace = go.Scatter3d(
        x=edge_x, y=edge_y, z=edge_z, mode='lines',
        line=dict(width=0.5, color='#888'),
        hoverinfo='none'
    )

    node_trace = go.Scatter3d(
        x=x, y=y, z=z, mode='markers',
        marker=dict(size=10, color='lightblue'),
        hoverinfo='text',
        text=[f"Node: {node}" for node in graph.nodes]
    )

    fig = go.Figure(data=[edge_trace, node_trace])
    fig.update_layout(title="3D Genomic Visualization", showlegend=False)
    fig.show()

visualize_interactive_3d(genome_graph)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import networkx as nx
import numpy as np

def visualize_static_3d_with_gradient(graph):
    # Generate 3D layout for the graph
    pos = nx.spring_layout(graph, dim=3, seed=42)
    x, y, z = zip(*[pos[node] for node in graph.nodes])
    
    # Map z-axis (depth) to a color gradient
    z_norm = (np.array(z) - min(z)) / (max(z) - min(z))  # Normalize z values for color mapping
    colors = plt.cm.viridis(z_norm)

    # Plot the edges
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    for edge in graph.edges():
        edge_x = [pos[edge[0]][0], pos[edge[1]][0]]
        edge_y = [pos[edge[0]][1], pos[edge[1]][1]]
        edge_z = [pos[edge[0]][2], pos[edge[1]][2]]
        ax.plot(edge_x, edge_y, edge_z, color='gray', linewidth=0.5, alpha=0.8)

    # Plot the nodes
    ax.scatter(x, y, z, c=colors, s=50, edgecolor='k', depthshade=True)

    # Add color bar for depth
    mappable = plt.cm.ScalarMappable(cmap='viridis')
    mappable.set_array(z_norm)
    cbar = fig.colorbar(mappable, ax=ax, shrink=0.5, aspect=10)
    cbar.set_label("Depth (Z Axis)", fontsize=12)

    # Add labels and title
    ax.set_title("Static 3D Visualization of Genomic Data with Depth Gradient", fontsize=14)
    ax.set_xlabel("X Axis")
    ax.set_ylabel("Y Axis")
    ax.set_zlabel("Z Axis (Depth)")

    # Save the plot
    output_path = "static_3d_dna_visualization.png"
    plt.savefig(output_path, dpi=300)
    plt.show()

    print(f"Static 3D DNA visualization saved as {output_path}.")

# Call the function with your graph
visualize_static_3d_with_gradient(genome_graph)
```

```python
import plotly.graph_objs as go
import networkx as nx

def interactive_3d_visualization(graph):
    pos = nx.spring_layout(graph, dim=3, seed=42)
    
    # Extract node positions
    x, y, z = zip(*[pos[node] for node in graph.nodes])
    depths = [data.get('depth', 0) for _, data in graph.nodes(data=True)]

    # Create edges for relationships
    edge_x, edge_y, edge_z = [], [], []
    for edge in graph.edges():
        edge_x += [pos[edge[0]][0], pos[edge[1]][0], None]
        edge_y += [pos[edge[0]][1], pos[edge[1]][1], None]
        edge_z += [pos[edge[0]][2], pos[edge[1]][2], None]

    # Edge trace
    edge_trace = go.Scatter3d(
        x=edge_x, y=edge_y, z=edge_z,
        mode='lines',
        line=dict(color='lightgray', width=0.5),
        hoverinfo='none'
    )

    # Node trace with depth-based gradient
    node_trace = go.Scatter3d(
        x=x, y=y, z=z,
        mode='markers',
        marker=dict(
            size=8,
            color=depths,  # Depth gradient
            colorscale='Viridis',
            colorbar=dict(title="Depth"),
            opacity=0.8
        ),
        hoverinfo='text',
        text=[f"Node: {node}<br>Depth: {data.get('depth', 'N/A')}" for node, data in graph.nodes(data=True)]
    )

    # Combine into a figure
    fig = go.Figure(data=[edge_trace, node_trace])
    fig.update_layout(
        title="Interactive 3D Visualization of Genomic Data",
        showlegend=False,
        scene=dict(
            xaxis_title='X Axis',
            yaxis_title='Y Axis',
            zaxis_title='Z Axis',
            xaxis=dict(showbackground=False),
            yaxis=dict(showbackground=False),
            zaxis=dict(showbackground=False),
        )
    )
    fig.show()

# Call the function with the graph
interactive_3d_visualization(genome_graph)
```
